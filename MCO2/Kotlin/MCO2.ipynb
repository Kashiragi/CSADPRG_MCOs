```kotlin
import java.time.temporal.ChronoUnit
val withNewCols = validYearsDf
    .add("CostSavings") {
        val approved = this["ApprovedBudgetForContract"] as Double
        val cost = this["ContractCost"] as Double
        (approved) - (cost)
    }
    .add("CompletionDelayDays") {
        val start = this["StartDate"] as LocalDate
        val end = this["ActualCompletionDate"] as LocalDate
        ChronoUnit.DAYS.between(start, end)
    }
val data = withNewCols
```

```kotlin
val df_1 = data
    .convert("ApprovedBudgetForContract","CostSavings","CompletionDelayDays").toDouble()
    .select("Region", "MainIsland", "ApprovedBudgetForContract", "CostSavings", "CompletionDelayDays")
var groupedByRegion = df_1
//    .groupBy{ it["Region"], it["MainIsland"]}
    .groupBy("Region", "MainIsland")
    .aggregate{
        sumOf { it["ApprovedBudgetForContract"]  as Double} into "TotalBudget"
        medianOf {  it["CostSavings"] as Double} into "MedianSavings"
        meanOf {  it["CompletionDelayDays"] as Double} into "AvgDelay"

    //percentage of projects with delays >30 days by Region and MainIsland
//        add("HighDelayPct") { row ->
//            val totalProjects = row.count()
//            val delayAbove30: Double = row.count { (it["CompletionDelayDays"] as Double) > 30}
//            (delayAbove30 / totalProjects) * 100.00
//        }
    }
    .add("EfficiencyScore") {
        val median = it["MedianSavings"] as Double
        val avg = it["AvgDelay"] as Double
        median/avg
    }

groupedByRegion
```
